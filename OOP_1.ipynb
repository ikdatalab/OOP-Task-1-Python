__Question 1:__ Fundamental Constructor & Method Logic
Create a class named `Student` with the following requirements:

 - Class attribute: `institution = "Board of Secondary Education Karachi"`

 - Constructor accepting `name` (string) and `marks` (a list of 3 subject scores out of 100).

 - Instance method `calculate_average()` returning the average score.

 - Instance method `get_grade()` returning:

   - `"A1"` if average ≥ 80

   - `"A"` if 70 ≤ average < 80

   - `"B"` if 60 ≤ average < 70

   - `"Fail"` if average < 60

```python
class Student:
    institution = "Board of Scondary Education Karachi"
    def __init__(self,name,marks = None):
        self.name = name
        self.marks = marks if marks is not None else []

    def calculate_average(self):
        if not self.marks:
            return 0.0
        return sum(self.marks) / len(self.marks)

    def get_grade(self):
        ave = self.calculate_average()
        if ave >= 80:
            return ("A1")
        elif ave>= 70:
            return ("A")
        elif ave >= 60:
            return ("B")
        else:
            return ("FAIL")


s1 = Student("umer",[60,80,90])
print (s1.institution)
print (s1.name)
print (s1.calculate_average())
print (s1.get_grade())
```

__Question 3:__ Utility Function with `@staticmethod`

Design an inventory helper class for a local store chain:

 - Create a class `RetailUtility`.

 - Add a `@staticmethod` named `calculate_gst(price, rate=18)` that calculates and returns the total price including 18% General Sales Tax (GST).

 - Add a `@staticmethod` named `format_pkr(amount)` that returns a formatted currency string (e.g., `12500` -> `"PKR 12,500.00"`).
 
 -  Demonstrate using both static methods without instantiating `RetailUtility`

```python
class RetailUtility:
    @staticmethod
    def calculate_gst(price, rate=18):
        return price + (price * rate / 100)

    @staticmethod
    def format_pkr(amount):
        return f"PKR {amount:,.2f}"


print(RetailUtility.calculate_gst(1000))
print(RetailUtility.format_pkr(12500))
```

__Question 4:__ Encapsulation & Abstraction in E-Commerce

Simulate an automated checkout system for a Pakistani logistics platform (e.g., TCS or Daraz Express):

 -  Create a class `OrderProcessing`.
 - Constructor inputs: `order_id`, `item_price`, `weight_kg`, and `destination_city`.
 - Implement a private helper method `_calculate_shipping()`:
   - Standard rate within Karachi: PKR 150 flat rate.
   - Rate for other cities (e.g., Lahore, Rawalpindi): PKR 250 base + PKR 50 per kg above 1 kg.
 - Implement a public method `generate_invoice()` that hides shipping formula complexity and returns a complete bill breakdown dictionary: `{"order_id": ..., "subtotal": ..., "shipping": ..., "grand_total": ...}.`

```python
class OrderProcessing:
    def __init__(self, order_id, item_price, weight_kg, destination_city):
        self.order_id = order_id
        self.item_price = item_price
        self.weight_kg = weight_kg
        self.destination_city = destination_city

    def _calculate_shipping(self):
        if self.destination_city.lower() == "karachi":
            return 150
        else:
             extra_weight = max(0,self.weight_kg-1)
             return (extra_weight * 50) + 250

    def generate_invoice(self):
        ship = self._calculate_shipping()
        return {"order_id":self.order_id,"subtotal": self.item_price, "shipping": ship , "grand_total": self.item_price + ship}

a1 = OrderProcessing(1212, 10000, 20,'karachi')
print(a1._calculate_shipping())
print(a1.generate_invoice())
```

__Question 5:__ Practical Coding Challenge (Account Ledger)

Build a `KhyberTextileAccount` class to handle client credit/debit transaction logs:

 - Attributes: `client_name`, `balance` (initial value defaults to `0.0`), and `history` (an empty list initialized in constructor).

 - Method `deposit(amount)`: Adds money to balance, records `"CREDIT: PKR <amount>"` in history with timestamp/date string, and returns updated balance.

 - Method `withdraw(amount)`: Deducts money if balance is sufficient; if insufficient, raises a custom message `"Insufficient Funds"` without modifying balance.
 
 - Method `statement()`: Prints the transaction ledger line-by-line along with the final balance.

```python
from datetime import datetime

class KhyberTextileAccount:
    def __init__(self,client_name, balance=0, history=None):
        self.client_name = client_name
        self.balance = balance
        self.history = history if history is not None else []

    def deposit(self,amount):
        self.amount = amount
        self.balance += self.amount
        formated = datetime.now().strftime("%B %d, %Y, %I:%M %p")
        a = f"CREDIT: PKR {self.amount} | {formated}"
        self.history.append(a)
        print("Updated balance:",self.balance, formated)

    def withdraw(self, amount):
        self.amount = amount
        if self.balance >= self.amount:
            self.balance -= self.amount
            b = f"DEBIT: PKR {self.amount} | {datetime.now().strftime("%B %d, %Y, %I:%M %p")}"
            self.history.append(b)
            print(f"your withdrawal {self.amount} successfully compeleted")
            print(f"your current balance is {self.balance}")
        else:
            print("Insufficient Funds")

    def statement(self):
        print(f"\n ---- {self.client_name} Transaction Statement ----")
        if not self.history:
            print("No transaction history yet")
        else:
            for record in self.history:
                print(f"• {record}\n")
            print(f"final account balance is PKR: {self.balance}")


a1 = KhyberTextileAccount("umer")
a1.deposit(10000)
a1.withdraw(1000)
a1.statement()
```

__Question 6:__ Real-World Scenario (Utility Bill Calculator)
Create a class `KElectricBilling` that calculates monthly electricity charges under local slab structures:

 - Class attributes: `meter_maintenance_fee = 150`

 - Constructor: `consumer_number`, `units_consumed`

 - Logic for `compute_bill()` based on slab rates:

   - First 100 units @ PKR 15/unit

   - Next 100 units (101-200) @ PKR 22/unit

   - Remaining units (above 200) @ PKR 32/unit

 - Return a summary object containing consumer number, total energy charges, maintenance fee, and net bill payable.

```python
class KElectricBilling:
    meter_maintenance_fee = 150
    def __init__(self, consumer_number, units_consumed):
        self.consumer_number = consumer_number
        self.units_consumed = units_consumed

    def compute_bill(self):
        total = 0
        unit = self.units_consumed
        if unit <= 100:
            total = unit * 15
        elif unit <= 200:
            total = (100 * 15) + ((unit - 100)*22)
        else:
            total = (100 * 15) + (100*22) + ((unit - 200)*32)

        net_bill = total + self.meter_maintenance_fee

        return {
        "consumer_number": self.consumer_number,
        "units_consumed": self.units_consumed,
        "energy_charges": total,
        "maintenance_fee": self.meter_maintenance_fee,
        "net_bill": net_bill,}
```

```python
c1 = KElectricBilling("123456",100)
c1.compute_bill()
```
